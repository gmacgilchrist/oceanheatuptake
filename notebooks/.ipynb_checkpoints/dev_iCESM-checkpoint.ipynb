```python
basedir = "/gws/nopw/j04/co2clim/datasets/iCESM1.2_Eocene/"
os.listdir(basedir)
```

```python
### PREPROCESS FUNCTION
def preprocess(ds):
    filename = ds.encoding["source"]
    filename_elements = filename.split(".")
    experiment = filename_elements[5]
    orbconfig = filename_elements[6]
    if orbconfig[:3]=="Orb":
        experiment += '.'+orbconfig
    timespan = filename_elements[-3]
    print(experiment,end=', ')
    print(timespan,end='\n')
    # place experiment as distinct dimension
    ds = ds.squeeze().expand_dims({'experiment':[experiment]})
    return ds
```

```python
filename = "b.e12.B1850C5CN.f19_g16.iPETM0[1369]*.nc"
paths = glob.glob(basedir+filename)
paths
```

```python
ds = xr.open_mfdataset(paths,preprocess=preprocess)
ds
```

# d18Osw : surface maps

```python
experiments = ["iPETM01x","iPETM03x","iPETM06x","iPETM09x"]
```

```python
ds["d18O"] = (((ds["R18O"]/0.9985)-1)*1e3)
```

```python
with ProgressBar():
    danow = ds["d18O"].mean("time").compute()
```

```python
cmap = cmocean.cm.balance
cmap.set_bad(color='darkgrey',alpha=0)
clims=[-2,2]
longname = r'$\delta^{18}O_{sw}$'
ne = len(experiments)
fig,axarr = plt.subplots(figsize=(9,4*ne),nrows=ne,subplot_kw={'projection':ccrs.Robinson(central_longitude=220)})
for i,experiment in enumerate(experiments):
    axnow = axarr[i]
    da = danow.isel(z_t=0).sel(experiment=experiment)
    # damask = np.ma.array(da,mask = np.isnan(da))
    im = axnow.pcolormesh(da['TLONG'],da['TLAT'],da,transform=ccrs.PlateCarree(),cmap=cmap)
    axnow.text(85,60,experiment,color='w',ha='left',fontweight=12,transform=ccrs.PlateCarree())
    im.set_clim(clims)
    axnow.gridlines(draw_labels=True,color='lightgrey',linestyle='--',alpha=0.5)
    if i==0:
        axnow.set_title(longname)
axc = fig.add_axes([0.92,0.4,0.02,0.2])
cb = fig.colorbar(im,cax=axc)
cb.set_label(longname)
```

# d18O sections

```python
## horizontal regrid
grid_1x1 = xr.Dataset()
grid_1x1['lon'] = xr.DataArray(data=0.5 + np.arange(360), dims=('x'))
grid_1x1['lat'] = xr.DataArray(data=0.5 -90 + np.arange(180), dims=('y'))
# initiate regridder
regrid_to_1x1 = xesmf.Regridder(ds.rename({'TLONG': 'lon',
                                           'TLAT': 'lat'}),
                                grid_1x1, 'bilinear', periodic=True)
# regrid dataset
ds_1x1 = regrid_to_1x1(ds).assign_coords({'x':grid_1x1['lon'],'y':grid_1x1['lat']})
```

```python
lon=339
anomaly = False
with ProgressBar():
    if anomaly:
        tmp = ds_1x1['d18O'].sel(x=lon,method='nearest').mean("time")
        dasec = tmp-tmp.sel(experiment="iPETM01x").compute()
    else:
        tmp = ds_1x1['d18O'].sel(x=lon,method='nearest').mean("time")
        dasec = tmp.compute()
```

```python
cmap = cmocean.cm.balance
cmap.set_bad(color='darkgrey',alpha=0.5)
clims=[-1,1]
longname = r'$\delta^{18}O_{sw}$'

ne = len(experiments)
fig,axarr = plt.subplots(figsize=(9,3*ne),nrows=ne)
for i,experiment in enumerate(experiments):
    axnow = axarr[i]
    da = dasec.sel(experiment=experiment)
    damask = np.ma.array(da,mask = np.isnan(da))
    im = axnow.pcolormesh(da['y'],da['z_t']*1e-2,damask,cmap=cmap)
    axnow.text(80,5200,experiment,color='k',ha='right',fontweight=12)
    axnow.invert_yaxis()
    axnow.set_ylabel(r'Depth [m]')
    if i==ne-1:
        axnow.set_xlabel(r'Latitude [$^\circ N$]')
    else:
        axnow.set_xticklabels([])
    im.set_clim(clims)
    if i==0:
        axnow.set_title(r' '+longname+', at '+str(lon)+' $^\circ E$')
axc = fig.add_axes([0.92,0.4,0.02,0.2])
cb = fig.colorbar(im,cax=axc)
cb.set_label(longname)
```
